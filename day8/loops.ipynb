1-times_table.py

```python
num = int(input("Which table? "))
end=int(input("upto which number"))
total=0
for i in range(1, end+1):
    print(f"{num} x {i} = {num * i}")
    total+=num*i
print("the sum of all products are",total)
# 5 x 1 = 5
# 5 x 2 = 10
# ...
# 5 x 10 = 50

# Your job:
# 1. ask "Up to?" and use it in range()
# 2. add up all the answers and print the total
```

02_guess_game.py


```python
secret = 7
tries = 0

while tries<3:
    guess = int(input("Guess (1 to 10): "))
    tries += 1

    if guess == secret:
        print(f"Correct! You took {tries} tries")
        break
    elif guess < secret:
        print("Too low")
    else:
        print("Too high")

# Your job:
# 1. give only 3 tries (hint: while tries < 3)
# 2. bonus: import random
#    secret = random.randint(1, 10)
```

03_atm.py


```python
balance = 1000
correct_pin = "1234"
tries = 0

while tries < 3:
    pin = input("Enter PIN: ")

    if pin == correct_pin:
        print("PIN correct")
        break
    else:
        tries += 1
        print("Wrong PIN")

if tries == 3: 
    print("Too many wrong attempts. Account locked.")
else:
    while True:
        print("1. Balance  2. Deposit  3. Withdraw  4. Exit")
        choice = input("Choose (1-4): ").strip()

        if choice == "1":
            print(f"Balance: Rs. {balance}")

        elif choice == "2":
            amount = int(input("Amount to deposit: "))

            # 2. Don't allow 0 or negative deposit
            if amount <= 0:
                print("Deposit must be greater than 0")
            else:
                balance += amount
                print(f"New balance: Rs. {balance}")

        elif choice == "3":
            amount = int(input("Amount to withdraw: "))

            if amount > balance:
                print("Not enough money")
            else:
                balance -= amount
                print(f"Take your cash. Left: Rs. {balance}")

        elif choice == "4":
            print("Thank you! Bye")
            break

        else:
            print("Please choose 1 to 4")
```

04_vowel_counter.py


```python
text = input("Type a sentence: ").lower()
vowels = 0
space=0
back=""
for letter in text:
    if letter in "aeiou":
        vowels += 1
    if " " in text:
        space+=1
    back=letter+back    
print(f"Vowels: {vowels}")
print(f"space: {space}")
print(f"printed backwards:{back}")

# Type a sentence: I love Nepal
# Vowels: 5

# Your job:
# 1. also count the spaces
# 2. print the sentence backwards with a loop
#    (hint: start with back = "", then back = letter + back)
```

05_marks_report.py

```python
marks = {"Ram": 78, "Sita": 92, "Hari": 35, "Gita": 64}
total = 0
top=0
topper=""
for name, mark in marks.items():
    if mark >= 40:
        print(f"{name}: {mark} Pass")
    else:
        print(f"{name}: {mark} Fail")
    total += mark

    if mark > top:
        top = mark
        topper = name

    if mark>80 :
        print("your grade is a")
    elif mark>70 :
            print("your grade is B")
    elif mark>60 :
            print("your grade is C")
average = total / len(marks)
print(f"Class average: {average}")
print(f"the topper of the class is {topper} with {top} marks")
# Your job:
# 1. find the topper (hint: top = 0, then check each mark)
# 2. give a grade A / B / C with elif (Day 07)
```

06_password_check.py

```python
has_number=False
has_capital = False
while True:
    password = input("New password: ")

    has_number = False
    for ch in password:
        if ch.isdigit():
            has_number = True
        if ch.isupper():
            has_capital= True
    if len(password) < 8:
        print("Too short. Use 8 or more letters")
    elif not has_number:
        print("Add at least one number")
    elif not has_capital:
        print("Add at least one capital character")
    else:
        print("Strong password. Saved!")
        break

# Your job: also ask for one capital letter
# (hint: ch.isupper())
```

07_fizzbuzz.py


```python
no_of_fizz=0
for n in range(1, 50):
    if n % 3 == 0 and n % 5 == 0:
        print("FizzBuzz")
    elif n % 3 == 0:
        print("Fizz")
        no_of_fizz+=1
    elif n % 5 == 0:
        print("Buzz")
    else:
        print(n)
print("The no of fizz is ",no_of_fizz)
# 1 2 Fizz 4 Buzz Fizz 7 8 Fizz Buzz 11 Fizz 13 14 FizzBuzz

# Your job:
# 1. go up to 50
# 2. count how many times Fizz was printed
```
