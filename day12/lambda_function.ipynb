```python
def add(a,b):
    return a+b

print(add(2,3))

addition= lambda a,b : a+b
print(addition(2,3))
```

```python
words=["apple","banana","fig"]
last= lambda w:w[-1]

for w in words:
    print(f"{w} -> {last(w)}")
```
